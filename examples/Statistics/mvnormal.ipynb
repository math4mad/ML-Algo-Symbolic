```python
from sympy.stats import Expectation, Normal, Probability, Poisson,MultivariateNormal, density, marginal_distribution,P,correlation,given
from sympy import symbols, Integral, Sum,MatrixSymbol,Matrix,simplify,expand,Eq
mu = symbols("mu")
mux=symbols("mu_x")
muy=symbols("mu_y")
rho=symbols("rho")
sigmax = symbols("sigma_x", positive=True)
sigmay = symbols("sigma_y", positive=True)
```

# Defing MvNormal  Symoblics  Formula

```python
## Define  Mean Vector
```

```python
mu=[mux,muy]
mu
```

##  Define Convariance matrix  of mvnormal

```python
Sigma=symbols("Sigma",positive=True)
Sigma=[[sigmax, rho*sigmax*sigmay], [rho*sigmax*sigmay, sigmay]]
#Matrix([[sigmax, rho*sigmax*sigmay], [rho*sigmax*sigmay, sigmay]])
Matrix(Sigma)
```

```python
m=MultivariateNormal('X', mu, Sigma)
x = symbols("x")
y = symbols("y")
density(m)

```

```python
simplify(density(m)(x,y))
```

```python
p=density(m)(x, y)

dist=given(p,Eq(y,y))

density(dist)
```
