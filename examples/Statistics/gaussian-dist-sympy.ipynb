```python
from sympy.stats import Expectation, Normal, Probability, Poisson,MultivariateNormal, density, marginal_distribution,P
from sympy import symbols, Integral, Sum,MatrixSymbol
mu = symbols("mu")
mu1=symbols("mu_x")
mu2=symbols("mu_y")
rho=symbols("rho")
sigmax = symbols("sigma_x", positive=True)
sigmay = symbols("sigma_y", positive=True)
#X = Normal("X", mu, sigma)
#Expectation(X)
#Expectation(X).evaluate_integral().simplify()
```

```python
m=MultivariateNormal('X', [mu1,mu2], [[sigmax, rho*sigmax*sigmay], [rho*sigmax*sigmay, sigmay]])
x = Symbol("x")
y = Symbol("y")
p=density(m)(x, y)
p
```
