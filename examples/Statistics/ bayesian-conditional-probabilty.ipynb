#  Bayesian Inference of  Normal Dist 

```python
from sympy.stats import Expectation, Normal, Probability, Poisson,MultivariateNormal, density, marginal_distribution,P,given
from sympy import symbols, Integral, Sum,MatrixSymbol,Matrix,simplify,expand,Eq
```

```python
x=symbols('x')
T = Normal('T', 1, 3) # Prior distribution
noise = Normal('eta', 0, 1.5)

#density(noise)
observation = T + noise

density(observation)  # 观测数据的密度函数
```

```python
data = [5 + noise,4+noise]
data
```

```python
T_posterior = given(T, Eq(observation,5))

density(T_posterior)
```

```python
pdf_prior = density(T)(x)
pdf_prior

```

```python
pdf_post=density(T_posterior)(x)
pdf_post
```

```python
import numpy as np
x_vals = np.linspace(-10, 10, 1000)

def  data(dist):
     return [dist.subs(x, val) for val in x_vals]

y_prior=[pdf_prior.subs(x, val) for val in x_vals]
y_post=[pdf_post.subs(x, val) for val in x_vals]

```

```python
import matplotlib.pyplot as plt 
plt.plot(x_vals, y_prior, label='Prior dist', color='blue')
plt.plot(x_vals, y_post, label='Posterior dist', color='red')
plt.title('Bayesian Inference')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()

plt.show()
```
